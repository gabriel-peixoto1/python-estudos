### **DICIONÁRIOS**

```python
# DICIONÁRIOS
# dicionários tem um conceito de chave/valor.
# associa um valor numérico ou não a uma outra coisa.

preços = {'lapis':5.50,'caneta':7.99,'caderno':45.00}

print(preços)
print(preços['lapis'])
print(preços['caneta'])
print(preços['caderno'])

# depois que um dicionário é definido, ele pode ser usado como definição no
# código inteiro, não apenas no bloco em que ele está.
```

```python

preços.keys()

# essa função é parecida com aquela type(x) do início.
# no momento que vc digita, ela retorna todas as chaves do dicionario citado.
```

```python
preços.values()

# essa aqui por sua vez retorna todos os valores do dicionario.
```

```python
print('lapis' in preços)
print('caderno' in preços)
print('K-med' in preços)

#fórmula para testar logicamente (T ou F) a existência de um elemento
```

```python
preços ['mochila'] = 150.0
preços ['agenda'] = 20.0
preços ['régua'] = 0.50

print(preços)

# fórmula para incluir um valor no dicionário
```

```python
del preços ['régua']

print(preços)

# fórmula para deletar um valor no dicionário
```

```python
print(preços)

# interessante que conforme são incluídos ou adicionados elementos, eles mudam
# e atualizam o dicionario dali pra frente.
```

```python
print(preços.get('caneta','NÃO ENCONTRADO!'))
print(preços.get('mochila','NÃO ENCONTRADO!'))
print(preços.get('quadro branco','NÃO ENCONTRADO!'))
print(preços.get('lapis','NÃO ENCONTRADO!'))
print(preços.get('regua','NÃO ENCONTRADO!'))
print(preços.get('seu cu','NÃO ENCONTRADO!'))
print(preços.get('caderno','NÃO ENCONTRADO!'))


# busca um valor, se não encontrar mostra segundo parametro
# só funciona um de cada vez. para encontrar vários, tem que usar o 'print'.
```

### **SETS**

```python
# SETS
# sets são bem parecidos com lista,
# mas SEM ORDENAÇÃO e SEM REPETIÇÃO de elementos.
# também são usadas chaves, como nos dicionários.

bichos = {'gato', 'cachorro', 'sapo', 'gato'}

print(bichos)

# repare que o output retorna em ordem aleatória e sem repetição.
```

```python
print ('gato' in bichos)
print('lapis' in bichos)
print('caderno' in bichos)
print('cachorro' in bichos)
```

```python
bichos.add('cavalo')
bichos.add('boi')
bichos.add('mãe')
bichos.add('rinoceronte')

print(bichos)
print(len(bichos))

# adicionar elemento
# usar a função len para contar os elementos
```

### **TUPLAS**

```python
# TUPLAS
# tuplas são LISTAS que NÃO PODEM SER ALTERADAS.
# tuplas são definidas entre parentêses.

tp1 = (1,2,3,4,5,6)

# não podem ter elementos adicionados, deletados ou alterados.
```

```python
print(3 in tp1)

print(len(tp1))
```
