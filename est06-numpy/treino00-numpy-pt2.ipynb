Montando uma matriz com valores aleatórios reais

```python
l1 = np.random.randn((3))
l2 = np.random.randn((3))
l3 = np.random.rand((3))

m1 = np.vstack([[l1],[l2],[l3]])

print(m1)

# usar a função vstack quando quiser unir vetores em linhas para formar matriz.
# usar a função hstack quando quiser unir vetores em colunas para formar matriz.
# se só usar o np.array, ele cria um bloco de matriz que, embora pareça uma 
# matriz, não vai funcionar como uma, e vai ferrar certas operações.
# a fórmula matriz.shape entrega esse erro: out = (a,b,c) ao invés de (a,b) 
# implica nesse problema.
```

Separando essa matriz por linhas e colunas:

```python
lin1 = m1[0,:]
lin2 = m1[1,:]
lin3 = m1[2,:]
col1 = m1[:,0]
col2 = m1[:,1]
col3 = m1[:,2]

dim = m1.shape
transp = m1.T

print(lin1)
print('-------------')
print(lin2)
print('-------------')
print(lin3)
print('-------------')
print(col1)
print('-------------')
print(col2)
print('-------------')
print(col3)
print('-------------')
print(dim)
print('-------------')
print(transp)
```

Criando uma matriz específica com um intervalo (reshape) + Transposta (t):

```python
m2 = np.arange(0,20).reshape((4,5))

print(m2)
print('-------------')

transposta = m2.T

print(transposta)

# reshape é um comamndo interessante para colocar uma série de números em 
# forma de uma matriz mais conveniente. 
```

Criando uma matriz com condição WHERE:

```python
m3 = np.random.randn(5,5)

print(m3)
print('-------------')

m4 = (m3>0)

print(m4)
print('-------------')

m5 = np.where(m4>0,1,0)

print(m5)

# associei o m4 a uma condição booleana que me retorna se minha afirmativa é 
# verdadeira ou falsa. 

# depois, a função where associa a condição verdadeira = 1 e falsa = 0.
```
