Pandas - Planilha - Data Frame

Resumão:

A biblioteca 'Pandas' possibilita diversas manipulações dentro de uma planilha (data frame).
A planilha de dados (csv) deve estar baixada no mesmo diretório (a esquerda) que o código está sendo utilizado.

Funções aprendidas:
1. dados.shape = saber as dimensões da tabela.
2. dados.describe = dados estatísticos dos números da tabela: média, máximo, mínimo, desvio padrão...
3. dados.head(n) = 'n' primeiras linhas da tabela.
4. dados.tail(n) = 'n' últimas linhas da tabela.
5. dados[['nome_coluna']] = filtra a tabela pela coluna.
6. dados.loc[[0:3]] = filta as primeiras quatro colunas. não precisa necessariamente ser um intervalo.
7. dados.iloc[a:b,x:y] = filtra da coluna 'a' até 'b' e da linha 'x' até 'y'.
8. dados.loc[dados['coluna'] == i] = filtra apenas os valores da coluna específica que atendem a condição "== i" (suporta < ou > também).
obs.: também é possível expandir essa forma para coletar dados de colunas específicas para linhas que atendem a condição pedida (ver minha_idade)
9. pd.Series = cria séries matemáticas de números ou strings e numera cada item. As séries podem ser criadas manualmente, com vetores (numpy) ou com o próprio data frame.
10. dados.rename = serve para renomear colunas.
11. dados.drop = serve para excluir colunas.
obs.: usar o comando inplace = true exclui/renomeia as colunas de forma permanente dali pra frente, ao invés de fazê-lo apenas para aquele bloco.
