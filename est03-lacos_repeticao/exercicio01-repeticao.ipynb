Calcular a média de um aluno que cursou a disciplina de Programação I, a partir da leitura das notas M1, M2 e M3; passando por um cálculo da média aritmética. Após a média calculada, devemos anunciar se o aluno foi aprovado, reprovado ou pegou exame
- Se a média estiver entre 0.0 e 4.0, o aluno está reprovado
- Se a média estiver entre 4.1 e 6.0, o aluno pegou exame
- Se a média for maior do que 6.0, o aluno está aprovado
- Se o aluno pegou exame, deve ser lida a nota do exame. Se a nota do exame for maior do que 6.0, está aprovado, senão; está reprovado

```python
m1 = float(input('digite a primeira nota: '))
m2 = float(input('digite a segunda nota: '))
m3 = float(input('digite a terceira nota: '))

media = (m1 + m2 + m3)/3

if m1 < 0 or m1 > 10:
  print('a primeira nota está errada.')
elif m2 < 0 or m2 > 10:
  print('a segunda nota está errada.')
elif m3 < 0 or m3 > 10:
  print('a terceira nota está errada.')

#ainda não está do jeito que eu quero. tem duas saídas e eu só queria uma.
#como fazer o código encerrar caso as primeiras condições não sejam atendidas?
#lembro de uma função 'end', mas não encontrei um jeito de aplicá-la aqui.

if media >= 0 and media <= 4:
  print('o aluno foi reprovado.')
elif media > 4 and media <= 6:
  print('o aluno pegou exame especial.')
elif media > 6 and media <= 10:
  print('o aluno foi aprovado.')
else:
  print('Média inválida.')
```
