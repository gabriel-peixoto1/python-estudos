1) Ler 5 notas e informar a média

```python
# USANDO O FOR

i = 1
nota = soma = media = 0

for i in range(1,6):
    nota = float(input('Insira a nota do mlk:'))
    soma += nota
    i += 1
media = soma/5
print("a média é:",media)

if media >= 6:
    print('o safado foi aprovado')
else: 
    print('reprovado hihihi')
```

```python
# USANDO O WHILE

i = 1
nota = soma = media = 0

while i < 6:
    nota = float(input('Insira a nota do mlk: '))
    soma = soma + nota
    i = i + 1

media = soma/5
print('a média é:', media)

if media >= 6:
    print('o safado foi aprovado.')
else: 
    print('reprovado, paizão.')

# i += 1 é a mesma coisa que i = i + 1
```

2) Imprimir a tabuada do número 3 (3 x 1 = 3 ; 3 x 10 = 30)

```python
# USANDO O WHILE

i = 0

while i <= 10:
    print('3 x',i,'=',3*i)
    i += 1

#easyyyy, fiz em 5min.
```

```python
# USANDO O FOR

for i in range(1,11):
    print('3 x',i,'=',3*i)
    i+=1
```

3) Faça um programa que peça uma nota, entre zero e dez. Mostre uma mensagem caso o valor seja inválido e continue pedindo até que o usuário informe um valor válido.

```python
nota = float(input('Insira a nota do aluno: '))

while nota < 0 or nota > 10:
    print('A nota deve ser de 0 à 10, amigo.')
    nota = float(input('Digite novamente: '))

if nota >=0 and nota <=10:
    print('Nota armazenada com sucesso.')
```

4) Faça um programa que leia um nome de usuário e a sua senha e não aceite a senha igual ao nome do usuário, mostrando uma mensagem de erro e voltando a pedir as informações. 

```python
nome = input('Insira o nome de usuário: ')
senha = input('Insira a senha: ')

while nome == senha:
    print('o nome de usuário não deve ser igual à senha.')
    nome = input('digite novamente o nome de usuário:')
    senha = input('digite novamente a senha')

if nome != senha:
    print('usuário e senha salvos com êxito.')
```
