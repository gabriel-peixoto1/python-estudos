```python
#vamos tentar brincar com laços de repetição.
#while crescente
count = 1
while count <= 5:
    print(count)
    count += 1
```

```python
#while decrescente
count = 12
while count >= 2:
    print(count)
    count -= 1.5
```

```python
#Escreva um programa em Python que:
#Use um laço for para percorrer os números de 1 a 10.
#Ignore (não imprima) os números ímpares, usando o comando continue.
#Imprima apenas os números pares.

for i in range(0,11,1):
    if i % 2 == 0 and i != 0:
        print(i)

#o % como sendo o módulo ou 'resto da divisão por' é muito útil para pares/ímpares.
#o range pega o primeiro número do intervalo, mas não o segundo (?)
```

```python
#Escreva um programa em Python que:
#Use um laço while para pedir ao usuário que digite números inteiros.
#Enquanto o número digitado for positivo, o programa deve continuar pedindo
#novos números. Quando o usuário digitar um número negativo, o laço deve ser
#interrompido usando o comando break. Ao final, exiba: "Programa encerrado."
#Dica: comece atribuindo um valor inicial à variável antes do while.

numero = 0
#iniciar definindo uma condição que será sempre verdade, para que o "while True"
#não pare de funcionar.

while True:
    numero = int(input('insira um número inteiro:'))

    if numero < 0:
        print('Programa encerrado, zé.')
        break

#no momento que o programa lê o 'break' ele para totalmente. interessante.
```
