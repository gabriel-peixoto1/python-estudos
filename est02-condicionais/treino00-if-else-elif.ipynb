```python
y = float(input('Sua nota, de 0 a 10, para o filme: '))

if y > 5 and y < 7:
  print('Esse filme é ok.')
elif y == 5:
  print('Esse filme é mediano.')
elif y < 5 and y > 2.5:
  print('Esse filme é ruim.')
elif y <= 2.5 and y >= 0:
  print('Esse filme é um lixo.')
elif y >= 7 and y <= 10:
  print('Esse filme é bom.')
else:
  print('A nota é de 0 a 10, otário.')

#o elif serve para que a primeira condição (do if) seja negada, mas não todas
#as outras (?), e no final, o else nega todas elas. de certa forma, o elif fica
#entre o if e o else, descrevendo outras condições.
```
