1) Escreva um programa em Python que:
- Use um laço for para percorrer os números de 1 a 20.
- Imprima cada número na tela.
- Quando o número for igual a 12, o programa deve interromper o laço usando break.

```python
for i in range(0,21,1):
    while i != 12:
        print(i)
        i += 1
    break

#sem querer criei um laço infinito nessa kkkk. quase explodiu o note.
#a relação "while/break" parece ser parecida com "if/else". um encerra o outro.
```

2) Escreva um programa em Python que:
- Use um laço while para percorrer os números de 1 a 10.
- Pule (não imprima) os números múltiplos de 3, usando o comando continue.
- Imprima apenas os números que não são múltiplos de 3.

```python
for i in range(1,11,1):
    while i%3 != 0:
        print(i)
        i += 1
    continue
```
