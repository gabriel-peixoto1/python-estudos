1) Faça um programa que leia 4 notas, mostre as notas e a média na tela.

```python
tds_notas = []

for i in range(4):
    nota = float(input('Insira a nota aqui: '))
    tds_notas.append(nota)

media = sum(tds_notas)/len(tds_notas)

print(tds_notas)
print('Média de notas:', media)

# a função LISTA.append(VALOR) integra um valor dentro de uma nota vazia.
# a função sum(LISTA) soma todos os valores dentro de uma lista.
```

2) Faça um programa que leia um vetor de 5 números inteiros, mostre a soma, a multiplicação e os números.

```python
numeros = []

for i in range(5):
    num = int(input('Insira um número inteiro: '))
    numeros.append(num)

resultado = 1 
for n in numeros:
    resultado = resultado * n

soma = sum(numeros)
mult = resultado

print('os números digitados são:', numeros)
print('a soma dos números é:', soma)
print('o produto entre os números é:', mult)

# foi interessante criar manualmente uma formula para produto de uma lista.
# essa forma de calcular nota usando o for e listas é bem melhor.  
```

3) Faça um Programa que leia um vetor A com 10 números inteiros, calcule e mostre a soma dos quadrados dos elementos do vetor.

```python
vetor = []

for i in range(10):
    numero = int(input('Insira um número inteiro: '))
    numero = numero * numero
    vetor.append(numero)

soma = sum(vetor)

print('Quadrados perfeitos:', vetor)
print('Soma dos quadrados:', soma)
```

4) Foram anotadas as idades e alturas de 06 alunos. Faça um Programa que determine quantos alunos com mais de 13 anos possuem altura inferior à média de altura desses alunos.

```python
idades = []
alturas = []

for _ in range(6):
    idd = int(input('Insira a idade do aluno: '))
    idades.append(idd)

for n in range(6):
    h = int(input('Insira a altura do aluno: '))
    alturas.append(h)

media_h = sum(alturas)/len(alturas)

procurados = []

for i in range(6):
    if idades[i] > 13 and alturas[i] < media_h:
        procurados.append(alturas[i])

print('A média de altura da turma é:', media_h)
print('Aluno classificado se idade < 13 anos e altura >', media_h,'cm.')
print('O número de alunos classificado é:',len(procurados),'.')

# puta que pariu, esse código ficou INCRÍVEL!
```

5) Dicionário: Crie um dicionário para armazenar o nome e a nota de 3 alunos, fazendo a leitura dos valores por meio de uma estrutura de repetição. Depois, crie uma nova estrutura de repetição para somar todas as notas e retornar a média

```python
notas = {'gabriel':10.0,'melissa':2.0,'jao':6.0}

media = sum(notas.values())/3

print('a média dos alunos é:', media, 'pontos.')
```
